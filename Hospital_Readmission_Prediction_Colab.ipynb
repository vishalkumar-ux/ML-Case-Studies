# Hospital Readmission Prediction

Logistic Regression with L2 Regularization

This notebook predicts whether a patient will be readmitted within 30 days.

## 1. Import libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

## 2. Create patient data

```python
data = {
    "diagnosis_code": [1, 2, 1, 3, 2, 3, 1, 2, 3, 1],
    "heart_rate": [80, 95, 88, 110, 90, 105, 78, 92, 115, 85],
    "blood_pressure": [120, 130, 125, 145, 135, 140, 118, 128, 150, 122],
    "prior_visits": [1, 3, 2, 5, 4, 6, 1, 3, 7, 2],
    "readmitted": [0, 1, 0, 1, 1, 1, 0, 1, 1, 0]
}

df = pd.DataFrame(data)
df
```

## 3. Separate input and output

```python
X = df.drop("readmitted", axis=1)
y = df["readmitted"]
```

## 4. Split the data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

## 5. Scale the data

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## 6. Create the Logistic Regression model

```python
model = LogisticRegression(
    penalty="l2",
    C=1.0,
    max_iter=1000
)
```

## 7. Train the model

```python
model.fit(X_train, y_train)
```

## 8. Get prediction probabilities

```python
y_prob = model.predict_proba(X_test)[:, 1]
y_prob
```

## 9. Evaluate using ROC-AUC

```python
auc = roc_auc_score(y_test, y_prob)
print("ROC-AUC:", auc)
```

## 10. Clinical cost of errors

A false negative happens when the model predicts low risk but the patient is actually readmitted. This can be serious because the patient may not receive extra follow-up or care.

A false positive happens when the model predicts high risk but the patient is not readmitted. This can lead to unnecessary monitoring and use of hospital resources.

In this case, false negatives are generally more costly than false positives, so a real clinical system should pay special attention to identifying high-risk patients.

This small dataset is only for demonstration. A real hospital model would need a much larger patient dataset, proper validation, and clinical testing.
